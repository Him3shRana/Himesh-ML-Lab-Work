### Here the Roll Number is Being Treated by Pandas as a Floating point value (Float 64)
#### So for Showing up the Roll numbers As it is we will Read Roll No. as a string, not a number

```python
import pandas as pd

df = pd.read_csv('../ML Data.csv', dtype={'ROLL NO.': str})
print(df.head())
```

That **leftmost column (0, 1, 2, 3, 4)** is not part of our actual data — it's not a column from our CSV. It's pandas' built-in row index/label, added automatically so every row has a unique identifier pandas can use internally (for selecting rows, sorting, merging, etc.).

Soooo If we want to use Roll No. as the index instead (common and often cleaner for this kind of data) Below Code having **inplace = True** is used

```python
df = pd.read_csv('../ML Data.csv', dtype={'ROLL NO.': str})
df.set_index('ROLL NO.', inplace=True)
print(df.head())
```

### Basic Exports

```python
# To CSV
df.to_csv('exported_data.csv', index=False)

# To Excel
df.to_excel('exported_data.xlsx', index=False)

# To JSON
df.to_json('exported_data.json', orient='records', indent=4)
```
